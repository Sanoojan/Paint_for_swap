```python

Intermediate_path="Save_intermediates/intermediate"
pred_x0_path="Save_intermediates/pred_x0"
src_path="dataset/FaceData/CelebAMask-HQ/Val"
target_path="dataset/FaceData/CelebAMask-HQ/Val_target"

select_intermediates=[0,30,50,65,70,75,77]
path_list=[src_path,target_path]

Image_no=60

Labels=["Source","Target","0","30","50","65","70","75","77"]

import os
import cv2
import matplotlib.pyplot as plt
import numpy as np
import natsort

# plot intermediate results
def read_images(path_list,Intermediate_path, pred_x0_path,select_intermediates,Image_no):
    images = []
    for path in path_list:
        filenames = os.listdir(path)
        filenames = natsort.natsorted(filenames)  # Sort filenames naturally
        selected_filenames = [filenames[Image_no]]
        images.append([cv2.cvtColor(cv2.imread(os.path.join(path, filename)), cv2.COLOR_BGR2RGB) for filename in selected_filenames])
    filenames = os.listdir(Intermediate_path)
    filenames = natsort.natsorted(filenames)  # Sort filenames naturally
    intermediate_folder_path=os.path.join(Intermediate_path,filenames[Image_no])
    filenames = os.listdir(intermediate_folder_path)
    filenames = natsort.natsorted(filenames)  # Sort filenames naturally
    selected_filenames = [filenames[i] for i in select_intermediates]
    print(selected_filenames)
    for i in range(len(selected_filenames)):
        images.append([cv2.cvtColor(cv2.imread(os.path.join(intermediate_folder_path, selected_filenames[i])), cv2.COLOR_BGR2RGB)])
    
    
    return images


images=read_images(path_list,Intermediate_path, pred_x0_path,select_intermediates,Image_no)

def visualize_images(images):
    num_paths = len(images)
    num_images_per_path = len(images[0])
    x_size=num_paths*6
    y_size=num_images_per_path*6
    fig, ax = plt.subplots(num_images_per_path, num_paths,figsize=(x_size, y_size))
    print(ax.shape)
    for i in range(num_images_per_path):
        for j in range(num_paths):
            ax[j].imshow(images[j][i])
            ax[j].axis('off')
            # set title
            # if i == 0:
            #     ax[j].set_title(Labels[j], fontsize=30)
            
            
    plt.subplots_adjust(wspace=0, hspace=0)
    # save plot
    # plt.savefig("Image_outputs/visualize_hair_swap.png")
    
    plt.show()

visualize_images(images)
```
