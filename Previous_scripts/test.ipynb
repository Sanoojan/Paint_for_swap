```python
import cv2
import numpy as np
import os

# Define the folder containing mask images and the output video path
input_folder = "results_video/An Old Man Doing Exercises For The Body And Mindtoyanlecun_swapmask_frames"
output_video = "output_video2.mp4"
fps = 10  # Frames per second

# Define colors for each mask value (0-11)
# Format: BGR (Blue, Green, Red)
colors = {
    0: (0, 0, 0),       # Background - Black
    1: (0, 0, 255),     # Lip - Red
    2: (0, 255, 0),     # Eyebrows - Green
    3: (255, 0, 0),     # Eyes - Blue
    4: (0, 255, 255),   # Hair - Yellow
    5: (255, 255, 0),   # Nose - Cyan
    6: (255, 0, 255),   # Skin - Magenta
    7: (128, 0, 128),   # Ears - Purple
    8: (128, 128, 0),   # Belowface - Olive
    9: (128, 128, 128), # Mouth - Gray
    10: (0, 128, 255),  # Eye Glass - Orange
    11: (255, 255, 255) # Ear Rings - White
}

# Get list of image files
image_files = sorted([os.path.join(input_folder, f) for f in os.listdir(input_folder) if f.endswith(('.png', '.jpg'))])

# Read the first image to get dimensions
first_image = cv2.imread(image_files[0], cv2.IMREAD_GRAYSCALE)
height, width = first_image.shape

# Create a VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
video_writer = cv2.VideoWriter(output_video, fourcc, fps, (width, height))

# Iterate through images and write to video
for image_file in image_files:
    # Read mask image
    mask = cv2.imread(image_file, cv2.IMREAD_GRAYSCALE)
    
    # Create a color image
    color_image = np.zeros((height, width, 3), dtype=np.uint8)
    
    # Assign colors based on mask values
    for value, color in colors.items():
        color_image[mask == value] = color
    
    # Write the frame to the video
    video_writer.write(color_image)

# Release the video writer
video_writer.release()
print(f"Video saved to {output_video}")
```
